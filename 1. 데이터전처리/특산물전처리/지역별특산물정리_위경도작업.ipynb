```python
import requests
import pandas as pd

KAKAO_API_KEY = 'e642fce799374dab2bff7d50a730f015'

def get_coordinates(place_name):
    url = "https://dapi.kakao.com/v2/local/search/keyword.json"
    headers = {
        "Authorization": f"KakaoAK {KAKAO_API_KEY}"
    }
    params = {
        "query": place_name
    }

    response = requests.get(url, headers=headers, params=params)
    if response.status_code == 200:
        data = response.json()
        if data['documents']:
            first_result = data['documents'][0]
            name = first_result['place_name']
            lat = first_result['y']
            lon = first_result['x']
            return name, lat, lon
        else:
            return place_name, None, None
    else:
        print(f"Error {response.status_code}: {response.text}")
        return place_name, None, None

# 🟡 위도/경도 결과 누적
results = []

for cheong in cheong_list:
    name, lat, lon = get_coordinates(cheong)
    results.append({
        '청명': cheong,
        '검색된_이름': name,
        '위도': lat,
        '경도': lon
    })

# ✅ DataFrame 생성
df_coords = pd.DataFrame(results)
```

```python
def merge_with_coords(df, df_coords):
    """
    df: 특산물 데이터 (청명 컬럼 포함)
    df_coords: 청명, 위도, 경도가 들어있는 좌표 DataFrame

    return: 좌표가 병합된 DataFrame
    """
    merged_df = df.merge(df_coords, on='청명', how='left')
    return merged_df
```
