```python
import numpy as np
import pandas as pd
```

```python
criteria = ['편의', '교통', '농업_지원', '농업_매물', '서비스', '교육']
importance_dict = {
    '은퇴생계형': [9, 7, 5, 3, 2, 1],
    '청년창농형': [5, 2, 3, 9, 7, 1],
    '가족정착형': [9, 1, 3, 5, 2, 7],
    '영농상속형': [9, 3, 7, 2, 5, 1]
}

# 결과를 담을 딕셔너리
weights_dict = {}

# 각 유형별로 쌍대비교→정규화→가중치 계산
for name, importance in importance_dict.items():
    n = len(criteria)
    P = np.ones((n, n))
    for i in range(n):
        for j in range(n):
            P[i, j] = importance[i] / importance[j]
    col_sums = P.sum(axis=0)
    N = P / col_sums  # 열 정규화
    w = N.mean(axis=1)  # 행 평균
    weights_dict[name] = w

# DataFrame으로 변환
df_weights = pd.DataFrame(weights_dict, index=criteria).T
df_weights.index.name = '유형'
```

```python
df_weights.to_csv('AHP_weights.csv',encoding='utf-8-sig')
```
