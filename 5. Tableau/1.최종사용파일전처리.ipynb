```python
import pandas as pd
```

```python
data=pd.read_csv("적용유형별_가중합_상위10.csv", encoding="utf-8-sig")
data.head()
```

# 그래프표시용 데이터 수정

```python
data.columns
```

```python
avg_columns = ['농업_지원', '농업_매물', '교통', '교육', '편의', '서비스']
temp = pd.DataFrame()
for col in avg_columns:
    temp_copy = data[["address","적용유형"]].copy()
    temp_copy['시설유형'] = col
    temp = pd.concat([temp, temp_copy], ignore_index=True)
temp=temp.sort_values(by=["address","적용유형","시설유형"]).reset_index(drop=True)
```

```python
temp["점수"]=pd.NA
temp.head()
```

```python
for i, j, k in zip(temp["address"], temp["적용유형"], temp["시설유형"]):
    temp.loc[(temp["address"]==i)&(temp["적용유형"]==j)&(temp["시설유형"]==k), "점수"] = data.loc[(data["address"]==i)&(data["적용유형"]==j), k].values[0]
```

```python
temp.to_csv("행정코드매칭데이터.csv", encoding="utf-8-sig", index=False)
```
